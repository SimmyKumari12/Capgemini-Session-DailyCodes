#### Which city had maximum revenue
OR
#### Which manager earned maximum revenue

```python
df.groupby("City")["Revenue"].sum().sort_values(ascending=False)
```

```python
df.groupby("Manager")["Revenue"].sum().sort_values(ascending=False)
```

#### Average Revenue Of November and December

```python
df[df["Date"].dt.month == 11]["Revenue"].mean()
```

```python
df[df["Date"].dt.month == 12]["Revenue"].mean()
```

#### Compare revenue by purchase type

```python
val = df.groupby("Purchase Type")["Revenue"].sum().sort_values(ascending=False).reset_index
```

```python
sns.barplot(x="Purchase Type",y = "Revenue", data = val).sum().values
plt.show()
```
