```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("/content/Diwali Sales Data.csv",encoding="unicode_escape")
df
```

```python
df.isnull().sum()
```

```python
df.drop(columns=['Status','unnamed1'],inplace = True)
```

```python
df.columns
```

```python
df.dropna(inplace=True)
```

```python
df["Amount"] = df["Amount"].astype("int")
```

```python
sns.countplot(x="Gender",data = df)
plt.title("Male vs Female Comparison")
plt.show()
```

```python
sales_gender = df.groupby("Gender", as_index=False)["Amount"].sum()

sns.barplot(x="Gender", y="Amount", data=sales_gender)

plt.title("Sales Gender Wise")
plt.show()
```

```python
age_order = ['0-17','18-25','26-35','36-45','46-50','51-55','55+']
sns.countplot(x="Age Group",hue="Gender",data = df, order=age_order)
plt.title("Age Group Comparison")
plt.show()
```

```python
sales_group = df.groupby("Age Group")["Amount"].sum().reset_index().sort_values(by="Age Group")
sns.barplot(x="Age Group",y="Amount",data = sales_group)
plt.title("Sales by age group comparison")
plt.show()
```

from the above graph , we can conclude that age group of 26-35 is doing the most amount of shopping

Total no of orders state wise

```python
ordersPerState = df.groupby("State")["Orders"].sum().sort_values(ascending = False).head(5).plot(kind="barh")
plt.title("Orders Per State")
plt.show()
```

```python
amountPerState = df.groupby("State")["Amount"].sum().sort_values(ascending = False).head(5).plot(kind="barh")
plt.title("Amount Per State")
plt.show()
```
